# **3. Data Preparation**

```python
"""## 3.1 Imputación de Valores Nulos"""
# Verificar valores nulos
print("Valores nulos antes del tratamiento:")
print(df.isnull().sum())

# Imputación de la edad con la mediana
if 'age' in df.columns:
    df['age'] = df['age'].fillna(df['age'].median())

# Imputación de embarked con la moda
if 'embarked' in df.columns:
    df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])

print("\nValores nulos después de la imputación:")
print(df.isnull().sum())


"""## 3.2 Limpieza de Datos Categóricos"""
# Revisar categorías existentes
if 'sex' in df.columns:
    print("\nCategorías de sex:")
    print(df['sex'].unique())

if 'embarked' in df.columns:
    print("\nCategorías de embarked:")
    print(df['embarked'].unique())

if 'class' in df.columns:
    print("\nCategorías de class:")
    print(df['class'].unique())

"""## 3.3 Transformación o Eliminación de Features"""
print("\nColumnas antes de la limpieza:")
print(df.columns.tolist())

# Eliminar variables con demasiados valores nulos o redundantes
columnas_eliminar = ['deck', 'alive']

df = df.drop(columns=columnas_eliminar, errors='ignore')

print("\nColumnas después de la limpieza:")
print(df.columns.tolist())

# Codificación de variables categóricas (One-Hot Encoding)
df = pd.get_dummies(df, drop_first=True)

print("\nDataset después de One-Hot Encoding:")
print(df.head())

# Verificación final
print("\nInformación final del dataset:")
df.info()


"""### Visualización de Valores Atípicos"""
fig, axes = plt.subplots(1, 2, figsize=(12,5))

if 'age' in df.columns:
    sns.boxplot(y=df['age'], ax=axes[0])
    axes[0].set_title('Boxplot Edad')

if 'fare' in df.columns:
    sns.boxplot(y=df['fare'], ax=axes[1])
    axes[1].set_title('Boxplot Tarifa')

plt.tight_layout()
plt.show()


print("\nDimensiones finales del dataset:")
print(df.shape)

df.head()
```

# **3).Data Preparation (Parte II)**



```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

"""### 3.1 Verificación final del dataset"""
print("Dimensiones del dataset:")
print(df.shape)

print("\nValores nulos:")
print(df.isnull().sum().sum())


"""### 3.2 Estandarización de variables numéricas"""

columnas_numericas = ['age', 'fare']

scaler = StandardScaler()

df[columnas_numericas] = scaler.fit_transform(df[columnas_numericas])

print("\nVariables estandarizadas:")
print(df[columnas_numericas].head())


"""### 3.3 División entrenamiento y prueba"""

X = df.drop('survived', axis=1)
y = df['survived']

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nConjunto de entrenamiento:")
print(X_train.shape)

print("\nConjunto de prueba:")
print(X_test.shape)

print("\nVariable objetivo:")
print(y.name)

print("\nCantidad de variables predictoras:")
print(X.shape[1])

print("\nPreparación completada correctamente.")
```

#4). Modeling:

Implementación y Entrenamiento: Entrenar y comparar al menos dos modelos supervisados ( Árboles de Decisión y Regresión Logistica)

```python
# Librerías de modelado
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (accuracy_score, confusion_matrix, classification_report, ConfusionMatrixDisplay, f1_score, precision_score, recall_score, roc_curve, auc)
import matplotlib.pyplot as plt
import numpy as np
from sklearn.ensemble import RandomForestClassifier
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled  = scaler.transform(X_test)
```

## 4.1). Modelo 1: Regresión Logística


```python
# Entrenamiento
lr = LogisticRegression(max_iter=1000, random_state=42)
lr.fit(X_train_scaled, y_train)

# Predicciones
y_pred_lr = lr.predict(X_test_scaled)

# Métricas
print("Regresión Logística")
print(f"Accuracy: {accuracy_score(y_test, y_pred_lr):.4f}")
print("\nReporte de Clasificación:")
print(classification_report(y_test, y_pred_lr, target_names=["No Sobrevivió", "Sobrevivió"]))
```

```python
cm_lr = confusion_matrix(y_test, y_pred_lr)
disp_lr = ConfusionMatrixDisplay(confusion_matrix=cm_lr, display_labels=["No Sobrevivió", "Sobrevivió"])
fig, ax = plt.subplots(figsize=(6, 5))
disp_lr.plot(ax=ax, colorbar=False, cmap="Blues")
ax.set_title("Matriz de Confusión – Regresión Logística")
plt.tight_layout()
plt.show()
```

## 4.2). Modelo 2: Árboles de Decisión

```python
# Entrenamiento
dt = DecisionTreeClassifier(max_depth=5, random_state=42)
dt.fit(X_train_scaled, y_train)

# Predicciones
y_pred_dt = dt.predict(X_test_scaled)

# Métricas
print("=== Árbol de Decisión ===")
print(f"Accuracy: {accuracy_score(y_test, y_pred_dt):.4f}")
print("\nReporte de Clasificación:")
print(classification_report(y_test, y_pred_dt, target_names=["No Sobrevivió", "Sobrevivió"]))
```

```python
cm_dt = confusion_matrix(y_test, y_pred_dt)
disp_dt = ConfusionMatrixDisplay(confusion_matrix=cm_dt, display_labels=["No Sobrevivió", "Sobrevivió"])
fig, ax = plt.subplots(figsize=(6, 5))
disp_dt.plot(ax=ax, colorbar=False, cmap="Greens")
ax.set_title("Matriz de Confusión – Árbol de Decisión")
plt.tight_layout()
plt.show()
```

## 4.3) Modelo 3: Random Forest

```python
# Entrenamiento
rf = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=42)
rf.fit(X_train_scaled, y_train)

# Predicciones
y_pred_rf = rf.predict(X_test_scaled)

# Métricas
print("=== Random Forest ===")
print(f"Accuracy: {accuracy_score(y_test, y_pred_rf):.4f}")
print("\nReporte de Clasificación:")
print(classification_report(y_test, y_pred_rf, target_names=["No Sobrevivió", "Sobrevivió"]))
```

```python
# Matriz de Confusión – Random Forest
cm_rf = confusion_matrix(y_test, y_pred_rf)
disp_rf = ConfusionMatrixDisplay(confusion_matrix=cm_rf, display_labels=["No Sobrevivió", "Sobrevivió"])
fig, ax = plt.subplots(figsize=(6, 5))
disp_rf.plot(ax=ax, colorbar=False, cmap="Oranges")
ax.set_title("Matriz de Confusión – Random Forest")
plt.tight_layout()
plt.show()
```

## 4.4). Compartivo entre Modelos


```python
# Comparación entre los modelos (Regresión Logística, Árbol de Decisión y Random Forest)
resultados = pd.DataFrame({
    "Modelo": ["Regresión Logística", "Árbol de Decisión", "Random Forest"],
    "Accuracy":  [accuracy_score(y_test, y_pred_lr),  accuracy_score(y_test, y_pred_dt), accuracy_score(y_test, y_pred_rf)],
    "Precision": [precision_score(y_test, y_pred_lr), precision_score(y_test, y_pred_dt), precision_score(y_test, y_pred_rf)],
    "Recall":    [recall_score(y_test, y_pred_lr), recall_score(y_test, y_pred_dt), recall_score(y_test, y_pred_rf)],
    "F1-Score":  [f1_score(y_test, y_pred_lr), f1_score(y_test, y_pred_dt), f1_score(y_test, y_pred_rf)],
})

resultados = resultados.set_index("Modelo").round(4)
print(resultados.to_string())
```

```python
# Gráfico de barras comparativo – 3 modelos
metricas = ["Accuracy", "Precision", "Recall", "F1-Score"]
x = np.arange(len(metricas))
ancho = 0.25

fig, ax = plt.subplots(figsize=(10, 5))
b1 = ax.bar(x - ancho, resultados.loc["Regresión Logística", metricas], ancho, label="Regresión Logística", color="steelblue")
b2 = ax.bar(x,         resultados.loc["Árbol de Decisión",   metricas], ancho, label="Árbol de Decisión",   color="seagreen")
b3 = ax.bar(x + ancho, resultados.loc["Random Forest",       metricas], ancho, label="Random Forest",       color="darkorange")

ax.set_xticks(x)
ax.set_xticklabels(metricas)
ax.set_ylim(0, 1.18)
ax.set_ylabel("Valor")
ax.set_title("Comparación de Métricas – 3 Modelos")
ax.legend()
for b in [b1, b2, b3]:
    ax.bar_label(b, fmt="%.3f", padding=2, fontsize=7)
plt.tight_layout()
plt.show()
```

```python
# Curva ROC – comparación de los 3 modelos
fig, ax = plt.subplots(figsize=(8, 6))

for modelo, y_score, color, nombre in [
    (lr,  lr.predict_proba(X_test_scaled)[:, 1],  "steelblue",  "Regresión Logística"),
    (dt,  dt.predict_proba(X_test_scaled)[:, 1],  "seagreen",   "Árbol de Decisión"),
    (rf,  rf.predict_proba(X_test_scaled)[:, 1],  "darkorange", "Random Forest"),
]:
    fpr, tpr, _ = roc_curve(y_test, y_score)
    roc_auc = auc(fpr, tpr)
    ax.plot(fpr, tpr, color=color, lw=2, label=f"{nombre} (AUC = {roc_auc:.3f})")

ax.plot([0, 1], [0, 1], 'k--', lw=1, label="Clasificador aleatorio")
ax.set_xlabel("Tasa de Falsos Positivos")
ax.set_ylabel("Tasa de Verdaderos Positivos")
ax.set_title("Curva ROC – Comparación de Modelos")
ax.legend(loc="lower right")
plt.tight_layout()
plt.show()
```

## 5.1). Elección Mejor modelo

Basándonos en los criterios de éxito definidos, el modelo elegido es la Regresión Logística. En problemas donde queremos un buen equilibrio entre minimizar los falsos positivos y los falsos negativos, el F1-Score es una métrica clave. La Regresión Logística no solo obtuvo el mejor Accuracy, sino también el mejor Recall y F1-Score para identificar la clase minoritaria (los que sobrevivieron), además de ser un modelo computacionalmente eficiente y altamente interpretable (podemos ver los pesos asignados a variables como el sexo o la clase).

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# 1. Extraemos los coeficientes REALES (sin valor absoluto)
coeficientes = lr.coef_[0]
nombres_variables = X.columns

# 2. Creamos un DataFrame y lo ordenamos de menor a mayor
df_coef = pd.DataFrame({'Variable': nombres_variables, 'Coeficiente': coeficientes})
df_coef = df_coef.sort_values(by='Coeficiente')

# 3. Asignamos colores: Rojo para negativos (izquierda), Verde para positivos (derecha)
colores = ['#d62728' if coef < 0 else '#2ca02c' for coef in df_coef['Coeficiente']]

# 4. Generamos la gráfica
plt.figure(figsize=(10, 6))
plt.title('Impacto de las Variables en la Supervivencia (Regresión Logística)')

# Dibujamos las barras
plt.barh(df_coef['Variable'], df_coef['Coeficiente'], color=colores)

# Dibujamos una línea negra en el 0 para separar positivos de negativos
plt.axvline(x=0, color='black', linewidth=1)

plt.xlabel('Impacto (Negativo = Disminuye supervivencia / Positivo = Aumenta supervivencia)')
plt.tight_layout()
plt.show()
```

### 5.2). Ética y Sesgos

El dataset del Titanic es un ejemplo clásico para estudiar el sesgo inherente en los datos históricos.

- Sesgo de Género y Edad: La política histórica de "mujeres y niños" en los botes salvavidas hace que las variables sex_female y age tengan un peso predictivo altísimo. El modelo aprenderá explícitamente a favorecer la supervivencia de mujeres y penalizar la de los hombres adultos.

- Sesgo Socioeconómico: La variable pclass (clase del pasajero) refleja directamente los ingresos y el estatus social. Los pasajeros de 1ra clase tenían sus cabinas más cerca de las cubiertas superiores y prioridad en el rescate.

- Implicación Ética: Si este modelo (o uno entrenado con datos que contengan este tipo de patrones) se utilizara hoy en día para priorizar recursos de rescate, otorgar seguros de vida o calcular primas de riesgo médico, sería profundamente discriminatorio y poco ético, ya que tomaría decisiones de vida o muerte basándose en el género y el poder adquisitivo de la persona.

La aplicación de la metodología CRISP-DM fue fundamental para el éxito de este proyecto. ### 6.2). Limitaciones del Modelo

A pesar de los buenos resultados, el enfoque presenta limitaciones:

- Pérdida de información: Al imputar la edad con la mediana global, ignoramos que la edad podría variar según el estatus (los de 1ra clase tendían a ser mayores).

- Ingeniería de características (Feature Engineering) limitada: No aprovechamos información valiosa que podría derivarse, como extraer los títulos de los nombres ("Mr.", "Dr.", "Miss") para estimar edades faltantes con mayor precisión.
